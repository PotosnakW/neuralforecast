```python
import numpy as np
import pandas as pd

from plot_utils import _evaluate_simglucose, _evaluate_ohiot1dm
from neuralforecast.losses.numpy import mae, rmse
```

```python
def get_dataset_results(results_dir, dataset_name, trials, model_name, approach, metric, values):
    metric_group = []
    for trial in trials:
        results = pd.read_csv(results_dir+f'/{dataset_name}_6/{approach}_models/trial_{approach}_{trial}/forecasts.csv')
        if 'ohiot1dm' in dataset_name:
            avmask = pd.read_csv(f'../datasets/ohiot1dm_exog_9_day_test_avmask.csv')
            metric_result = _evaluate_ohiot1dm(results, avmask, model_name, metric, values)
        elif 'simglucose' in dataset_name:
            metric_result = _evaluate_simglucose(results, model_name, metric, values)
        metric_group.append(metric_result)
    
    return np.mean(np.array(metric_group)), np.std(np.array(metric_group))
```

## Baseline results

```python
model_names = ['AutoNHITS', 'AutoNBEATSx', 'AutoTFT']
metric = 'mae' #'rmse'
trials = range(8)
dataset_name = 'ohiot1dm' #'simglucose'
approach = 'baseline'
results_dir = '../results/multivariate_models'

mean_results_df1 = pd.DataFrame()
sd_results_df1 = pd.DataFrame()
for model_name in model_names:
    mm, ms = get_dataset_results(results_dir, dataset_name, trials, model_name, approach, metric, values='all')
    final_mean = pd.DataFrame([[mm]], index=[model_name]).round(3)
    final_sd = pd.DataFrame([[ms]], index=[model_name]).round(3)
    mean_results_df1 = pd.concat([mean_results_df1, final_mean], axis=0)
    sd_results_df1 = pd.concat([sd_results_df1, final_sd], axis=0)

mean_results_df2 = pd.DataFrame()
sd_results_df2 = pd.DataFrame()
for model_name in model_names:
    mm, ms = get_dataset_results(results_dir, dataset_name+'_exog', trials, model_name, approach, metric, values='all')
    final_mean = pd.DataFrame([[mm]], index=[model_name]).round(3)
    final_sd = pd.DataFrame([[ms]], index=[model_name]).round(3)
    mean_results_df2 = pd.concat([mean_results_df2, final_mean], axis=0)
    sd_results_df2 = pd.concat([sd_results_df2, final_sd], axis=0)

mean_results_df3 = pd.DataFrame()
sd_results_df3 = pd.DataFrame()
for model_name in model_names:
    mm, ms = get_dataset_results(results_dir, dataset_name, trials, model_name, approach, metric, values='critical')
    final_mean = pd.DataFrame([[mm]], index=[model_name]).round(3)
    final_sd = pd.DataFrame([[ms]], index=[model_name]).round(3)
    mean_results_df3 = pd.concat([mean_results_df3, final_mean], axis=0)
    sd_results_df3 = pd.concat([sd_results_df3, final_sd], axis=0)

mean_results_df4 = pd.DataFrame()
sd_results_df4 = pd.DataFrame()
for model_name in model_names:
    mm, ms = get_dataset_results(results_dir, dataset_name+'_exog', trials, model_name, approach, metric, values='critical')
    final_mean = pd.DataFrame([[mm]], index=[model_name]).round(3)
    final_sd = pd.DataFrame([[ms]], index=[model_name]).round(3)
    mean_results_df4 = pd.concat([mean_results_df4, final_mean], axis=0)
    sd_results_df4 = pd.concat([sd_results_df4, final_sd], axis=0)

print(pd.concat([
           mean_results_df1,
           mean_results_df2,
           mean_results_df3,
           mean_results_df4,
          ], axis=1)
     )
print(pd.concat([
           sd_results_df1,
           sd_results_df2,
           sd_results_df3,
           sd_results_df4,
          ], axis=1)
     )
```
