```python
def get_perc_diff(result_baseline, results_new):
    perc_mean = np.mean(100*(np.mean(result_baseline, axis=0) - np.mean(results_new, axis=0))/np.mean(result_baseline, axis=0))
    perc_max = np.max(100*(np.mean(result_baseline, axis=0) - np.mean(results_new, axis=0))/np.mean(result_baseline, axis=0))
    return perc_mean, perc_max

def get_pval(result_baseline, results_new):
    t_stat, p_value = ttest_rel(np.mean(results_new, axis=0), np.mean(result_baseline, axis=0), alternative='less')
    return p_value
```
