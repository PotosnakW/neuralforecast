```python
import pandas as pd
import torch
from computational_complexity_utils import *
```

```python
dataset_name = 'simglucose' #'ohiot1dm' #
paths = [ 
    f'../results/{dataset_name}_6/baseline_models/trial_baseline_0/autonhits_0.ckpt',
    f'../results/{dataset_name}_exog_6/baseline_models/trial_baseline_0/autonhits_0.ckpt',
    f'../results/{dataset_name}_exog_6/sum_total_models/trial_sum_total_0/autonhitstreat_0.ckpt',
    f'../results/{dataset_name}_exog_6/treat_models/trial_treat_0/autonhitstreat_0.ckpt',
    
    f'../results/{dataset_name}_6/baseline_models/trial_baseline_0/autonbeatsx_0.ckpt',
    f'../results/{dataset_name}_exog_6/baseline_models/trial_baseline_0/autonbeatsx_0.ckpt',
    f'../results/{dataset_name}_exog_6/sum_total_models/trial_sum_total_0/autonbeatsxtreat_0.ckpt',
    f'../results/{dataset_name}_exog_6/treat_models/trial_treat_0/autonbeatsxtreat_0.ckpt',
    
    f'../results/{dataset_name}_6/baseline_models/trial_baseline_0/autotft_0.ckpt',
    f'../results/{dataset_name}_exog_6/baseline_models/trial_baseline_0/autotft_0.ckpt',
    f'../results/{dataset_name}_exog_6/sum_total_models/trial_sum_total_0/autotft_0.ckpt',
    f'../results/{dataset_name}_exog_6/treat_models/trial_treat_0/autotft_0.ckpt',
]
dataset_names = [
    dataset_name,
    dataset_name+'_exog',
    dataset_name+'_exog',
    dataset_name+'_exog',
    dataset_name,
    dataset_name+'_exog',
    dataset_name+'_exog',
    dataset_name+'_exog',
    dataset_name,
    dataset_name+'_exog',
    dataset_name+'_exog',
    dataset_name+'_exog',
]

names = [
    'NHITS',
    'NHITS Exog.',
    'NHITS Sum-Total',
    'NHITS PK',
    'NBEATSx',
    'NBEATSx Exog.',
    'NBEATSx Sum-Total',
    'NBEATSx PK',
    'TFT',
    'TFT Exog.',
    'TFT Sum-Total',
    'TFT PK',
]

model_flop_count = {}
model_num_param = {}
inference_time = {}
mem = {}
for path, dataset_name, name in zip(paths, dataset_names, names):
    print(name)
    checkpoint = torch.load(path)
    state_dict = checkpoint['state_dict']
    hparams = checkpoint['hyper_parameters']
    
    model_name = path.split('/')[-1]
    model_name = model_name.split('_')[0]
    model = get_model(model_name, dataset_name, hparams)
    inp = get_inp(dataset_name, model_name)
    model.load_state_dict(state_dict)
    
    model_flop_count[name] = get_flops(model, inp)
    model_num_param[name] = count_parameters(model)
    inference_time[name] = get_inference_time(model, inp=inp)
    mem[name] = get_mem(model, inp)
```
