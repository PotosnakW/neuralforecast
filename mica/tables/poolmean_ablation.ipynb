# Mean-pool memory ablation (MAE)

Each gate next to its mean-pool counterpart, where the memory is the channel average $A = \frac{1}{C}\sum_c V^{(c)}$ and does not depend on the query. Blue = the query-dependent memory beats mean-pool. Reads `results_csv/*/results.csv` (built by `build_results.py`); writes to `TABLES_DIR`.

```python
from table_utils import *
```

```python
COLUMNS = []
for bb in ['moment', 'patchtst']:
    COLUMNS += [
        (f'{bb}_vanilla', bb),
        (f'{bb}_shared', f'{bb}_shared_incl'),
        (f'{bb}_shared_pm', f'{bb}_shared_pm'),
        (f'{bb}_layerwise', f'{bb}_layerwise_incl'),
        (f'{bb}_layerwise_pm', f'{bb}_layerwise_pm'),
        (f'{bb}_mica', f'{bb}_mlpquery_incl'),
        (f'{bb}_mica_pm', f'{bb}_mlpquery_pm'),
    ]
IMPROVEMENT_PAIRS = [(f'{bb}_{g}_pm', f'{bb}_{g}')
                     for bb in ['moment', 'patchtst'] for g in ['shared', 'layerwise', 'mica']]
```

```python
table, means = make_table(COLUMNS, 'mae', improvement_pairs=IMPROVEMENT_PAIRS)
table
```
