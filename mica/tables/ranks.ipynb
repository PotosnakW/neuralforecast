# Average model ranks

Mean rank per model (1 = best). Only datasets where every model in the comparison has a result are ranked; the dropped ones are printed.

```python
from table_utils import *
```

```python
MAIN = [
    ('MOMENT', 'moment'), ('PatchTST', 'patchtst'),
    ('MOMENT-MICA', 'moment_mlpquery_incl'), ('PatchTST-MICA', 'patchtst_mlpquery_incl'),
    ('iTransformer', 'itransformer'), ('iTransformer-T5', 'itransformer_t5'),
    ('TimerXL', 'timerxl'), ('Crossformer', 'crossformer'), ('TSMixer', 'tsmixer'),
    ('MLP', 'mlp'), ('Chronos-2', 'chronos2'), ('Toto-2', 'toto2'), ('TimesFM-3', 'timesfm3'),
]
```

```python
for metric in ['mae', 'rmse']:
    print(f'--- main models, {metric.upper()}')
    means, _, _ = collect(MAIN, metric)
    display(average_ranks(means))
```
